### Exercise 1

Given two floats `a` and `b`, and some tolerance `tol`, write an expression that will test whether `a` and `b` are within `tol` of each other.

```python
from math import isclose
a = 0.1 + 0.1 + 0.1
b = 0.3

print(a, b, a == b)

# do not compare float numbers with equality sign '=='

# using isclose function from math
print(isclose(a, b))

# using built-in absolute function
print(abs(a - b))
print(abs(a - b) < 1e-3)
```

### Exercise 2

Assume you have some variable `elapsed` that contains elapsed time in seconds.

Create three new variables: `hours`, `minutes` and `seconds`, that represent the number of hours, minutes and seconds represented by `elapsed`.

For example, if `elapsed = 7835`, then `hours = 2`, `minutes = 10` and `seconds = 35`

```python
from math import floor
# seconds
elapsed = 7835

# / 60 minutes, / 60 hours
hours_complete = elapsed / 60 / 60
print(f'Hours Complete: {hours_complete}')

hours = floor(hours_complete)
print(f'Hours: {hours}')

rest_minutes = hours_complete - hours
print(f'Rest Minutes: {rest_minutes}')

minutes_complete = rest_minutes * 60
print(f'Minutes Complete: {minutes_complete}')

minutes = floor(minutes_complete)
print(f'Minutes: {minutes}')

rest_seconds = minutes_complete - minutes
print(f'Rest Seconds: {rest_seconds}')

seconds_complete = rest_seconds * 60
print(f'Seconds Complete: {seconds_complete}')

seconds = floor(seconds_complete)
print(f'Seconds: {seconds}')

print(f'Time: {hours}:{minutes}:{seconds}')
```
