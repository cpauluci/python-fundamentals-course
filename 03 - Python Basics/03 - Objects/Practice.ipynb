Everthing in python is object. integer, float, boolean, strings...
Class is a blueprint of the object attributes (variables - state and methods - functionality)

```python
class Account:
    # special method called constructor
    def __init__(self, account_number, account_type, initial_balance):
        self.account_number = account_number
        self.account_type = account_type
        self.balance = initial_balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f'Deposited {amount}')
            print(f'New balance is: {self.balance}')
        else:
            print(f'{amount} is invalid.')

    def withdraw(self, amount):
        if amount > 0 and amount <= self.balance:
            self.balance -= amount
            print(f'Withdrawn: {amount}')
            print(f'New balance: {self.balance}')
        else:
            if amount < 0:
                print(f'{amount} is invalid.')
            else:
                print('Insufficient funds.')
                print(f'Current balance is {self.balance}')
```

```python
my_account = Account('1234-567', 'travel', 1_000.00)
```

```python
my_account.account_number
```
